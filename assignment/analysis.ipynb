```python
import pandas as pd
from IPython.display import HTML, display
```

## Dados gerais


```python
def display_general_data(data_frame: pd.DataFrame):
    display(HTML(f"<strong>Features:</strong> {len(data_frame.columns)}"))
    display(HTML(f"<strong>Rows:</strong> {len(data_frame)}"))

    display(HTML("<strong>Missing values:</strong>"))
    with pd.option_context("display.max_rows", None):
        display(data_frame.isna().sum())

    display(HTML(f"<strong>Duplicated rows:</strong> {data_frame.duplicated().sum()}"))

    with pd.option_context("display.max_rows", None):
        display(data_frame.dtypes)
```

### Genis


```python
display_general_data(merged_source_genis_df)
```

### Rosids


```python
display_general_data(source_rosids_df)
```
