# Descrição das características


- `flow_bytes_per_second` has 3 infinite values, and 272 NaN. They are caused by a division by 0 when the duration of the flow is 0. When `total_forward_packet_length` is not 0, the `flow_bytes_per_second` becomes infinite; when it is, the result is NaN.

These are some transformations needed to make on ROSIDS dataset.

- `destination_port` feature must be categorized in ranges likewise GENIS. Port 0 must be stored as not_applicable. Port 1900 must be assigned to SSDP. 5353 into mDNS. 11311 into ROS.
- `source_port` must follow the same transformations as destination_port with some additions. Port 23 will be categorized as telnet.
- `protocol` feature has categorical data 0, 6, 17. It should be converted into None, TCP, UDP.
- `flow_bytes_per_second` has empty and infinite values, which should be transformed into pandas NaN.
- `forward_push_flag_count` must be removed, since its value is 0 for every entry.
- `forward_urgent_flag_count` must be removed, since its value is 0 for every entry.
- `backward_urgent_flag_count` must be removed, since its value is 0 for every entry.
- `urg_flag_count` must be removed, since its value is 0 for every entry.
- `cwe_flag_count` must be removed, since its value is 0 for every entry.
- `ece_flag_count` must be removed, since its value is 0 for every entry.
- `forward_bytes_per_block_average` must be removed, since its value is 0 for every entry.
- `forward_packets_per_block_average` must be removed, since its value is 0 for every entry.
- `forward_block_rate_average` must be removed, since its value is 0 for every entry.
- `backward_bytes_per_block_average` must be removed, since its value is 0 for every entry.
- `backward_packets_per_block_average` must be removed, since its value is 0 for every entry.
- `backward_block_rate_average` must be removed, since its value is 0 for every entry.
- `initial_forward_window_bytes` must be removed, since its value is -1 for every entry.
- `initial_backward_window_bytes` has -1 as sentinel value, which must be transformed into NaN.
- `forward_segment_size_minimum` must be removed, since its value is 0 for every entry.


```python
target_feature_specification = SOURCE_ROSIDS_SCHEMA.get_feature("label")


def display_rosids_feature(
    feature_specification: FeatureSpec,
    frequency_axis_scale: FrequencyAxisScale = FrequencyAxisScale.CUBE_ROOT,
    numerical_axis_scale: NumericalAxisScale = NumericalAxisScale.LOG1P,
):
    display_feature(
        feature_specification,
        target_feature_specification,
        source_rosids_df,
        frequency_axis_scale=frequency_axis_scale,
        numerical_axis_scale=numerical_axis_scale,
    )
```
